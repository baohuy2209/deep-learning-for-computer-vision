# 11. Variant V1: Local Multi-Instance Learning (MIL) Stream

This notebook demonstrates the fine-grained lesion branch:
- Native-resolution sliding-window patch extractor ($128 \times 128$)
- Instance backbone: EfficientNet-B0 ($D_{\text{local}} = 1280 \rightarrow 512$)
- Gated Attention Dual Pooling ($\mathbf{a}_m = \text{softmax}(\mathbf{w}^T (\tanh(\mathbf{V}\mathbf{h}_m) \odot \text{sigm}(\mathbf{U}\mathbf{h}_m)))$)


```python
import torch
from drdg.models.factory import build_variant

model = build_variant("v1", pretrained=False)
dummy_global = torch.randn(2, 3, 512, 512)
dummy_patches = torch.randn(2, 12, 3, 128, 128)
dummy_mask = torch.ones(2, 12)

out = model(global_img=dummy_global, patch_imgs=dummy_patches, patch_mask=dummy_mask)
print("V1 Logits shape:", out["logits"].shape)
print("Local bag attention weights shape:", out.get("attention_weights", "N/A"))
```

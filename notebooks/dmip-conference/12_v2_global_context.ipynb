# 12. Variant V2: Global Relational Context Stream

This notebook demonstrates the macro-structural context branch:
- Convolutional Block Attention Module (CBAM: Channel + Spatial Attention)
- Non-Local Self-Attention Block (Long-range spatial dependencies across retinal quadrants)
- Learned Quadrant Positional Tokens ($T_{\text{ST}}, T_{\text{SN}}, T_{\text{IT}}, T_{\text{IN}}$)


```python
import torch
from drdg.models.factory import build_variant

model = build_variant("v2", pretrained=False)
dummy_global = torch.randn(2, 3, 512, 512)
dummy_patches = torch.randn(2, 12, 3, 128, 128)
dummy_mask = torch.ones(2, 12)

out = model(global_img=dummy_global, patch_imgs=dummy_patches, patch_mask=dummy_mask)
print("V2 Logits shape:", out["logits"].shape)
print("Fused feature representation shape:", out.get("fused_features", "N/A"))
```

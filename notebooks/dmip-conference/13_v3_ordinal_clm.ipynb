# 13. Variant V3: Ordinal Cumulative Link Model (CLM)

This notebook demonstrates the proposed clinical ordinal regression head:
- Cumulative Link Model with complementary log-log (clog-log) link function
- Monotonic cutpoint parameterization via Softplus offsets ($b_k = b_{k-1} + \text{softplus}(\delta_k)$)
- Continuous Quadratic Weighted Kappa (QWK) surrogate training loss


```python
import torch
from drdg.models.factory import build_variant
from drdg.losses.qwk import ContinuousQWKLoss

model = build_variant("v3", pretrained=False)
dummy_global = torch.randn(4, 3, 512, 512)
dummy_patches = torch.randn(4, 16, 3, 128, 128)
dummy_mask = torch.ones(4, 16)

out = model(global_img=dummy_global, patch_imgs=dummy_patches, patch_mask=dummy_mask)
probs = out["probs"]
cutpoints = out.get("cutpoints")

print("V3 Probabilities (normalized [B, 5]):\n", probs)
print("Monotonic Cutpoints:", cutpoints)

# Verify probability mass conservation
sums = probs.sum(dim=-1)
print("Class probability sums (must be 1.0):", sums)
```

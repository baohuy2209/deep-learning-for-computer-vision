# 22. Grade-Specific Clinical Performance and Adjacent Error Analysis

This notebook evaluates per-grade sensitivity ($G_0$ to $G_4$), within-1 accuracy ($|y - \hat{y}| \le 1$), and severe classification errors ($|y - \hat{y}| \ge 2$).


```python
import numpy as np
import pandas as pd
from drdg.evaluation.grade_metrics import calculate_per_grade_metrics
from drdg.evaluation.ordinal_metrics import calculate_ordinal_metrics

# Synthesize paired ground-truth and prediction distribution
np.random.seed(42)
y_true = np.random.randint(0, 5, size=200)
# Add small Gaussian noise to simulate ordered predictions
y_pred = np.clip(np.round(y_true + np.random.normal(0, 0.5, size=200)), 0, 4).astype(int)

grade_res = calculate_per_grade_metrics(y_true, y_pred)
ordinal_res = calculate_ordinal_metrics(y_true, y_pred)

print(f"Exact Accuracy: {grade_res['exact_accuracy']:.4f}")
print(f"Within-1 Grade Accuracy: {grade_res['within_1_accuracy']:.4f}")
print(f"Severe Error Rate (|err| >= 2): {ordinal_res['severe_error_rate_ge_2']:.4f}")
pd.DataFrame(grade_res['per_grade']).T
```

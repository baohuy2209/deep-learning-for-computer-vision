# 10. Variant V0: Global Context Baseline (ResNet-50)

This notebook demonstrates the standard clinical baseline architecture:
- Backbone: Standardized ResNet-50 representation ($D_{\text{global}} = 2048 \rightarrow 512$)
- Pooling: Global Average Pooling (GAP)
- Output Head: Multi-class Softmax Cross-Entropy ($K = 5$)


```python
import torch
from drdg.models.variants.v0 import V0GlobalBaseline
from drdg.models.factory import build_variant

model = build_variant("v0", pretrained=False)
dummy_img = torch.randn(2, 3, 512, 512)
out = model(dummy_img)

print("V0 Forward Output keys:", list(out.keys()))
print("Logits shape:", out["logits"].shape)
print("Probs shape:", out["probs"].shape)
print("Predictions:", out["preds"])
```
